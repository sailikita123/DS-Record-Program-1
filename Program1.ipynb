```python
import pandas as pd
from docx import Document
from PyPDF2 import PdfReader
from PIL import Image
from pydub import AudioSegment
import cv2
```

```python
print("Reading CSV File")

url = "https://raw.githubusercontent.com/mwaskom/seaborn-data/master/iris.csv"

student_data = pd.read_csv(url)

print(student_data.head())

student_data.to_csv("iris_data.csv", index=False)

print("CSV file created successfully.")
```

```python
print("\nReading JSON Data")

json_data = student_data.to_json(orient="records", indent=4)

with open("iris_data.json", "w") as file:
    file.write(json_data)

new_data = pd.read_json("iris_data.json")

print(new_data.head())

print("JSON file created successfully.")
```

```python
print("\nReading Excel File")

student_data.to_excel("iris_data.xlsx", index=False)

excel_data = pd.read_excel("iris_data.xlsx")

print(excel_data.head())

print("Excel file created successfully.")
```

```python
print("\nReading PDF File")

reader = PdfReader("sample.pdf")

print("Total Pages :", len(reader.pages))

print(reader.pages[0].extract_text())
```

```python
from PIL import Image, ImageDraw

# Create a simple dummy JPEG image
width, height = 200, 100
img = Image.new('RGB', (width, height), color = 'yellow')

d = ImageDraw.Draw(img)
d.text((10,10), "Dummy JPEG", fill=(0,0,0))

img.save("sample.jpg")
print("Dummy 'sample.jpg' created.")
```

```python
print("\nReading Image")

picture = Image.open("sample.jpg")

print("Format :", picture.format)
print("Size :", picture.size)
print("Mode :", picture.mode)
```

```python
from pydub import AudioSegment

# Create a simple dummy audio file (silent for 1 second)
silence = AudioSegment.silent(duration=1000)  # 1 second of silence
silence.export("sample.wav", format="wav")
print("Dummy 'sample.wav' created.")
```

```python
print("\nReading Audio")

sound = AudioSegment.from_file("sample.wav")

print("Duration :", len(sound)/1000, "seconds")
print("Channels :", sound.channels)
print("Frame Rate :", sound.frame_rate)
```

```python
print("\nReading Video")

movie = cv2.VideoCapture("sample.mp4")

width = int(movie.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(movie.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = movie.get(cv2.CAP_PROP_FPS)

print("Width :", width)
print("Height :", height)
print("FPS :", fps)

movie.release()

print("\nExperiment Completed Successfully.")
```
